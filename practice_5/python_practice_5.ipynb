5. (9б). Створіть клас 3D-фігура без атрибутів, з методом get_volume для отримання об'єму фігури, що повертає 0 та __add __, який приймає self та other в якості аргументів, а повертає суму об'ємів 3d-фігур self та other.

```python
class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()
```

6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math
class Cylinder(Figure3D):
  def __init__(self, radius, height):
    self.radius = radius
    self.height = height
  def get_volume(self):
     return math.pi * self.radius**2 * self.height
class Cube(Figure3D):
  def __init__(self, side):
    self.side = side
  def get_volume(self):
     return self.side**3
cylinder = Cylinder(3,5)
cube = Cube(5)
total = cylinder + cube
print(total)
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
  def __init__(self, balance, id):
    self.__balance = balance
    self.__id = id
  @property
  def id(self):
      return self.__id
  @property
  def balance(self):
      return self.__balance
  @balance.setter
  def balance(self, value):
      if value < 0: print("Operation canceled. The balance cannot be negative.")
      else:
        self.__balance = value
  def deposit(self, amount):
    self.balance += amount
  def withdraw(self, amount):
    if self.balance - amount < 0: print("Withdrawal canceled. The balance cannot be negative.")
    else: self.balance -= amount

eWallet = EWallet(100, 123456)
print(eWallet.id)
print(eWallet.balance)
eWallet.deposit(199)
print(eWallet.balance)
eWallet.withdraw(123)
print(eWallet.balance)
eWallet.withdraw(1234567)
print(eWallet.balance)
eWallet.balance = -6
print(eWallet.balance)
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def __str__(self):
        return f"{self.name}: {self.price} грн"
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add_item(self, item):
        self.items.append(item)
        print(f"Item ‘{item.name}’ added to cart.")
    def remove_item(self, item_name):
        for item in self.items:
            if item.name == item_name:
                self.items.remove(item)
                print(f"Item ‘{item_name}’ has been removed from the cart.")
                return
        print(f"The item ‘{item_name}’ was not found in the shopping cart.")
    def show_items(self):
        if not self.items:
            print("The cart is empty.")
        else:
            print("Shopping cart contents:")
            for item in self.items:
                print(f"- {item}")
apple = Item("Apple", 20)
milk = Item("Milk", 45)
bread = Item("Bread", 30)
cart = ShoppingCart()
cart.add_item(apple)
cart.add_item(milk)
cart.add_item(bread)
cart.show_items()
cart.remove_item("Milk")
cart.show_items()
```

10. (13б). Створіть клас Vehicle з атрибутами brand (manufacturer) та production_age. Створіть ще один клас Engine з такими атрибутами, як type (ex. petrol) та size (volume, litres). Створіть клас Car, який успадковує обидва класи Vehicle та Engine. Продемонструйте використання множинної спадковості, створивши обʼєкт класу Car та отримавши mro для цього класу.

```python
class Vehicle:
    def __init__(self, brand, production_age, **kwargs):
        super().__init__(**kwargs)
        self.brand = brand
        self.production_age = production_age
class Engine:
    def __init__(self, type, size, **kwargs):
        super().__init__(**kwargs)
        self.type = type
        self.size = size
class Car(Vehicle, Engine):
    def __init__(self, brand, production_age, type, size):
        super().__init__(
            brand=brand,
            production_age=production_age,
            type=type,
            size=size
        )
    def info(self):
        return f"Auto: {self.brand}, Age: {self.production_age}, Engine: {self.type} {self.size}"
my_car = Car(brand="Toyota", production_age=2, type="Petrol", size=2.5)
print(my_car.info())
print(Car.mro())
```

Ітерабельний об'єкт - це об'єкт, по якому можна пройтися по елементах (наприклад, рядок, список, словник), а ітератор це об'єкт, який по одному видає елементи з ітерабельного об'єкта, він запам'ятовує, на якому елементі він зараз, і при виклику next() повертає наступний елемент.

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekDays:
    week = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
    def __init__(self):
        self.current_day = -1
    def __iter__(self):
        return self
    def __next__(self):
        if self.current_day < 6:
            self.current_day += 1
            return self.week[self.current_day]
        else:
            raise StopIteration
weekDays = WeekDays()
for day in weekDays:
  print(day)
```
